```python
import plotly.express as px
import pandas as pd
import os
import sqlite3


#seleione della directory corretta
cur_dir = os.getcwd()
csv_dir = os.path.join(cur_dir,"csv")

#connessione al db
conn  = sqlite3.connect("lavoro.db")

#estrapolazione dei 3 csv 
df_incidenza_spese = pd.read_csv(os.path.join(csv_dir,"incidenza_spese.csv"))
df_partecipazione_mercato_lavoro = pd.read_csv(os.path.join(csv_dir,"partecipazione_mercato_lavoro.csv"))
df_tasso_sppravvivenza_imprese = pd.read_csv(os.path.join(csv_dir,"tasso_sppravvivenza_imprese.csv"))

#creazione del df delle regioni 
df_regioni = pd.read_sql_query("SELECT * FROM regioni",conn)

#merge con la tabella regioni al fine di avere un grande df in cui sinao presenti sia i nomi delle regioni che l'indicazione dell'area geografica
df_incidenza_spese = df_incidenza_spese.merge(df_regioni[['Area_geografica','Regione']],left_on='Regione',right_on='Regione',how='inner')
df_partecipazione_mercato_lavoro = df_partecipazione_mercato_lavoro.merge(df_regioni[['Area_geografica','Regione']],left_on='Regione',right_on='Regione',how='inner')
df_tasso_sppravvivenza_imprese = df_tasso_sppravvivenza_imprese.merge(df_regioni[['Area_geografica','Regione']],left_on='Regione',right_on='Regione',how='inner')

display(df_incidenza_spese)
```

Andiamo a creare die grafici con i raggruppamenti estratti dai dati sopra analizzati

```python
fig = px.bar(
    df_incidenza_spese,
    x='Anno',
    y='Percentuale_spesa_imprese',
    color='Area_geografica',
    barmode='group'
)

fig.show()
```

```python
fig = px.treemap(
    df_incidenza_spese,
    values='Percentuale_spesa_imprese',
    path=['Area_geografica','Regione'],
    labels={'Percentuale_spesa_imprese':'Percentuale spesa imprese'}
)

fig.update_traces(textinfo='label+value',)

fig.show()
```

```python
fig = px.density_heatmap(
    df_incidenza_spese,
    x='Anno',
    y='Regione',
    z='Percentuale_spesa_imprese',
    histfunc='avg'
)

fig.show()
```
